```python
# إنشاء tuple
colors = ("أحمر", "أخضر", "أزرق", "أصفر")

# الوصول إلى العنصر الأول (الفهرس 0)
print(colors[0])  # أحمر

# الوصول إلى العنصر الثالث (الفهرس 2)
print(colors[2])  # أزرق

# الوصول إلى العنصر الأخير
print(colors[-1])  # أصفر

# slicing لاستخراج جزء من الـ tuple
print(colors[1:3])  # ('أخضر', 'أزرق')
```
