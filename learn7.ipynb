```python
def recusive_function(n):
  if n==0:
    return " تم انهاء العد"
  else:
    print(f"القيمة الحالية n: {n}")
    return recusive_function(n-1)
result=recusive_function(10)
print(result)
```

```python
def factorial(n):
  if n==0 or n==1 :
    return 1
  else :
    return n* factorial(n-1)
result=factorial(10)
print(result)
```

```python
def fibonacci(n) :
  if n==0 :
    return 0
  if n==1 :
    return 1
  else :
    return fibonacci(n-1)+fibonacci(n-2)
result=fibonacci(6)
print(result)
```

```python
def countdown(n) :
  print('start counter down !')
  while n>0:
    yield n
    n-=1
  print("stop counter")
counter=countdown(5)
print( next (counter))
print( next (counter))
print( next (counter))
print( next (counter))
print( next (counter))
print( next (counter))
```

```python
def m(n):
  result=[]
  for i in range(n):
    result.append(i*i)
  return result
num=10
b=m(num)
print(b)
```

```python
def b(n):
  for i in range(n):
    yield i*i
num=10
k=b(num)
print(k)
```

```python
def fibonacci(n) :
  a , b = 0 , 1
  count = 0
  while count < n :
    yield a
    a , b= b , a + b
    count += 1
for num in fibonacci(10) :
  print(  num,  end="")
```
