```python
accounts=[]
se_opr=0
while se_opr !='exit'and se_opr!="4":
    print('''اختر واحدة :
      1-create account
      2-login
      3-show accounts
      4-exit''')
    se_opr=input("select operation:")
    match se_opr:
        case"create account" |"1":
            username=input("ادخل اسم المستخدم:")
            password=input("ادخل كبمة المرور:")

            account={"username":username,"password":password}
            accounts.append(account)
            print("تم إنشاء الحساب بنجاح 🎉")
        case"login"|"2":
             username=input("ادخل اسم المستخدم:")
             password=input("ادخل كلمة المرور:")
             found=False
             for account in accounts:
                 if username == account.get("username") and password==account.get("password"):
                     found=True
                     break
             if found:
                print("تم تسجيل الدخول بنجاح 🔑✅")
             else :
              print("اسم المستخدم او كلمة المرور خطا الرجاء المحاولة مرة اخرى❌❌❌")
        case "show accounts"|"3":
            for account in accounts:
                print(account.items())
        case _:
            print("تم اغلاق البرنامج")
print("تم اغلاق البرنامج")

```
