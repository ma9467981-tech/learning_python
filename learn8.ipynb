```python
def simple() :
  yield 1
  yield 2
gen = simple()
try:
  print(next(gen))
  print(next(gen))
  print(next(gen))
except StopIteration :
  print("stoped")
```

```python
M_list=["Apple","panana",'orange']
M_iter=iter(M_list)
frist=next(M_iter)
print(frist)
second=next(M_iter)
print(second)
third=next(M_iter)
print(third)
```

```python
M_list=["Apple","panana",'orange']
M_iter=iter(M_list)
while True :
  try :
    fruit=next(M_iter)
    print(fruit)
  except StopIteration:
    print('stopIteration')
    break
```
