```python
def smart_decorator(func):
  def warpper(*args,**kwargs):
    print(f"{func.__name__}جاري استدعاء الدالة   ")
    result=func(*args,**kwargs)
    print("تم الانتهاء من التنفيذ!")
    return result
  return warpper
@smart_decorator
def greet(name,age):
  return f"اهلا:{name} عمرك {age} سنة"
@smart_decorator
def add_num(a,b):
  return a+b
m=greet("ahmed",23)
print(m)
add_num(a=int(input("a=")),b=int(input("b=")))
```
