```python
from datetime import datetime , timedelta
today=datetime.now()
print(today.strftime('%Y-%m-%d'))
next_week=today + timedelta(days=7)
print(next_week.strftime('%Y-%m-%d'))
m=today - timedelta(days=5)
print(m.strftime('%d-%m-%Y %H:%M'))
date1=datetime(2023,9,20)
date2=datetime(2026,9,25)
diff=date2 - date1
print(diff)
```
